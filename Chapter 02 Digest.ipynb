### Scaler Multiplication

 When you multiple a vector by a scaler constant $s$, the length of the vector will also be \|$\vec{c}$\|.s, as shown in the example below the orginal vector $\vec{c}$ has length of $\sqrt2$, 3*$\vec{c}$ has new vector of (3,3), which has the length of $\sqrt18$=3*$\sqrt2$

```python
draw(Arrow((1,1),(0,1),color='blue'),Arrow((1,2),(1,1),color='blue'),Arrow((1,2),(0,1),color='blue'),
        Arrow((3,1),(0,1),color='green'),Arrow((3,4),(3,1),color='green'),Arrow((3,4),(0,1),color='green')
     )
```

We are going to look at the how scalar multiplication can be used to scale drawings

```python
dino_vectors = [(6,4), (3,1), (1,2), (-1,5), (-2,5), (-3,4), (-4,4),
    (-5,3), (-5,2), (-2,2), (-5,1), (-4,0), (-2,1), (-1,0), (0,-3),
    (-1,-4), (1,-4), (2,-3), (1,-2), (3,-1), (5,1)
]

s=2

s_dinao_vectors = [scale(s, v) for v in dino_vectors] #list comprehension to scale the dino_vectors by a factor of s


#*dino_vectors unpacks the list of vectors into individual arguments for the Polygon function
draw(
Polygon(*dino_vectors, color=gray),
Polygon(*s_dinao_vectors, color=red)
)

```

### Polor Coordinates to Cartesian coordinates

We can see from the image below that all the Cartesian Coordinates can be also expressed by an angle from its orgin we call $\theta$ and length $r$ of the vector from its origin.

<img src="ch2_polar.png">

<img src="ch2_transformation.png">

```python
from math import cos, sin, pi
r=5
angle = 45/180*pi  # radians as input to the cos and sin functions 
v=to_cartesian((r, angle)) #take distance r and angle in radian to convert to cartesian coordinates
x=v[0] #x coordinate of the vector
y=v[1] #y coordinate of the vector
print(v)



draw(Points(*[(r*cos(theta/180*pi), r*sin(theta/180*pi)) for theta in range(0, 360, 5)]),
     Segment((0,0),(r*cos(angle), r*sin(angle)),color='red'),
     Arrow((x,0),(0,0),color='blue'), Arrow((x,y),(x,0),color='blue')
)

```

### Rotation of drawing by using Polar  Cartesian Coordinates

```python
dino_vectors = [(6,4), (3,1), (1,2), (-1,5), (-2,5), (-3,4), (-4,4),
    (-5,3), (-5,2), (-2,2), (-5,1), (-4,0), (-2,1), (-1,0), (0,-3),
    (-1,-4), (1,-4), (2,-3), (1,-2), (3,-1), (5,1)
]

rotation_angle = pi
dino_polar = [to_polar(v) for v in dino_vectors]
dino_rotated_polar = [(l,angle + rotation_angle) for l,angle in dino_polar]
dino_rotated = [to_cartesian(p) for p in dino_rotated_polar]
draw(
Polygon(*dino_vectors, color=gray),
Polygon(*dino_rotated, color=red)
)
```
